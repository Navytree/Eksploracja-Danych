Lista `dlugosc_zycia` zawiera oczekiwaną długość życia (w latach) w 14 krajach.

1. Napisz funkcję `srednia(liczby)`, która zwraca średnią arytmetyczną — **bez** używania `sum()` i NumPy (użyj pętli).
2. Napisz funkcję `mediana(liczby)`, która zwraca medianę. Pamiętaj o przypadku parzystej i nieparzystej liczby elementów. Wskazówka: `sorted()`.
3. Porównaj wyniki z `np.mean()` i `np.median()`.

🔵 **Dla ambitnych:** niech obie funkcje zgłaszają czytelny błąd (`raise ValueError(...)`), gdy lista jest pusta.

````python
dlugosc_zycia = [81.8, 81.0, 82.6, 83.0, 81.8, 84.3, 82.8, 81.9, 82.2, 82.0, 82.7, 79.8, 83.3, 82.0]
````

````python
def srednia(liczby):
    # TODO
    if len(liczby) == 0:
      raise ValueError("lista jest pusta")
    wynik = 0
    for l in liczby:
        wynik += l
    return wynik / len(liczby)


def mediana(liczby):
    # TODO
    if len(liczby) == 0:
        raise ValueError("lista jest pusta")
    liczby = sorted(liczby)
    if len(liczby) % 2 == 1:
        dlugosc_zycia.sort()
        return liczby[len(liczby) // 2]
    else:
        return (liczby[len(liczby) // 2 - 1] + liczby[len(liczby) // 2]) / 2

print("moja średnia:", srednia(dlugosc_zycia), " NumPy:", np.mean(dlugosc_zycia))
print("moja mediana:", mediana(dlugosc_zycia), " NumPy:", np.median(dlugosc_zycia))
````

### 🟢 Zadanie 2: zliczanie

Napisz funkcję `zlicz(lista)`, która zwraca słownik: element → liczba jego wystąpień.

Przykład: `zlicz(["a", "b", "a", "c", "a"])` → `{"a": 3, "b": 1, "c": 1}`

Sprawdź ją na liście `oceny`. Która ocena występuje najczęściej?

````python
oceny = [4, 3, 5, 4, 4, 2, 3, 5, 4, 3, 4, 5, 3, 4]

def zlicz(lista):
    # TODO
    dictionary = {}
    for l in lista:
      if l not in dictionary:
        amount = lista.count(l)
        dictionary[l] = amount
    return dictionary

d = zlicz(oceny)
dictionary = dict(sorted(d.items()))
print(dictionary)

print("Najczęściej występuje ocena:", max(dictionary, key=dictionary.get))
````

### Lista ≠ tablica

Te same operatory działają inaczej na listach i na tablicach NumPy:

````python
lista = [1, 2, 3]
tablica = np.array([1, 2, 3])

print("lista * 2   =", lista * 2)      # powtórzenie listy!
print("tablica * 2 =", tablica * 2)    # mnożenie każdego elementu
print("lista + lista     =", lista + lista)        # sklejenie list
print("tablica + tablica =", tablica + tablica)    # dodawanie element po elemencie
````

### 🟢 Zadanie 4: dlaczego NumPy?

Zsumuj liczby od 1 do 1 000 000 na **cztery sposoby** i zmierz czas każdego z nich magiczną komendą `%timeit`:

1. pętla `for`,
2. `sum()` na liście zbudowanej przez list comprehension,
3. `sum(range(...))`,
4. NumPy: `np.arange(...).sum()`.

Sprawdź, czy wszystkie dają ten sam wynik. Który sposób jest najszybszy i **dlaczego**? Odpowiedz w komórce tekstowej.

````python
from numpy.random.mtrand import f
N = 1000000

def suma_petla(n):
    # TODO
    wynik = 0
    for x in range(1,n+1):
        wynik += x
    return wynik


def suma_comprehension(n):
    # TODO
    return sum([x for x in range(1, n + 1)])

def suma_range(n):
    # TODO
    return sum(range(n+1))

def suma_numpy(n):
    # TODO
    return (np.arange(1, n+1)).sum()

print("Wynik petla:", suma_petla(N))
print("Wynik comprehension:", suma_comprehension(N))
print("Wynik range:", suma_range(N))
print("Wynik numpy:", suma_numpy(N))
wszystkie_rowne = (suma_petla(N) == suma_comprehension(N) == suma_range(N) == suma_numpy(N))
print("Czy są równe:", "Tak" if wszystkie_rowne else "Nie")
# TODO: sprawdź, że wyniki są równe, a potem zmierz czasy, np.:
# %timeit suma_petla(N)
%timeit suma_petla(N)
%timeit suma_comprehension(N)
%timeit suma_range(N)
%timeit suma_numpy(N)
````

Najszybszy jest sposób z użyciem biblioteki NumPy, ponieważ:
* W macierzy mamy ten sam typ danych, a w pętlach przy dodawaniu mamy dodatkową operację sprawdzania czy mamy ten sam typ danych
* Operacje NumPy są napisane w skompilowanym C, a nie interpretowanym Py, przez co są szybsze
* Dane z tablicy są w jednym bloku pamięci, dzięki czemu procesor może czytać całe kawałki danych na raz

### 🟢 Zadanie 5: operacje na tablicach

Dana jest tablica `x` z pomiarami temperatury (°C).

1. Policz średnią, odchylenie standardowe, minimum i maksimum.
2. Wybierz pomiary wyższe od średniej. Ile ich jest?
3. **Normalizacja min–max:** przeskaluj `x` do przedziału $[0, 1]$: $\;x' = \dfrac{x - x_{\min}}{x_{\max} - x_{\min}}$
4. **Standaryzacja:** $\;z = \dfrac{x - \bar{x}}{s}$. Sprawdź, że wynik ma średnią ≈ 0 i odchylenie standardowe ≈ 1.

Wszystko bez pętli!

````python
x = np.array([12.5, 14.0, 9.8, 17.3, 15.1, 11.2, 18.6, 16.0, 13.4, 10.9])
# TODO

srednia = x.mean()
odchylenie = x.std()
minimum = x.min()
maksimum = x.max()
print("Średnia:", srednia)
print("Odchylenie standardowe:", odchylenie)
print("Min, max:", minimum, maksimum)

wyzsze = x[x > srednia]
print("\nPomiary wyższe od średniej:", wyzsze)
print("Ile ich jest:", len(wyzsze))

normalizacja = (x - minimum) / (maksimum - minimum)
print("\nNormalizacja:", normalizacja)

standaryzacja = (x - srednia) / odchylenie
print("Standaryzacja:", standaryzacja)

print("Średnia po standaryzacji:", round(standaryzacja.mean(), 10))
print("Odchylenie po standaryzacji:", standaryzacja.std())


````

### 🔵 Zadanie 11 (dla ambitnych): rozkład normalny

1. Napisz funkcję gęstości rozkładu normalnego:
$$ f(x; \mu, \sigma) = \frac{1}{\sigma\sqrt{2\pi}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right) $$
2. Dla trzech par $(\mu, \sigma)$: (0, 1), (0, 2), (2, 1) wygeneruj po 10 000 liczb (`rng.normal`) i narysuj ich histogram (`density=True`) razem z krzywą gęstości policzoną ze wzoru.
3. Trzy wykresy jeden pod drugim, każdy z tytułem zawierającym $\mu$ i $\sigma$, opisami osi i legendą.

````python
# TODO
def gestosc(x, mu, sigma):
    wspolczynnik = 1 / (sigma * np.sqrt(2 * np.pi))
    wykladnik = -((x - mu) ** 2) / (2 * (sigma ** 2))
    return wspolczynnik * np.exp(wykladnik)

rng = np.random.default_rng(42)
pary = [ (0, 1), (0, 2), (2, 1) ]

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8, 10))
osie = [ax1, ax2, ax3]

for ax, (mu, sigma) in zip(osie, pary):
    dane = rng.normal(loc=mu, scale=sigma, size=10000)
    ax.hist(dane, bins=50, density=True, alpha=0.6, color="teal", edgecolor="white", label="Histogram danych")

    x = np.linspace(mu - 4*sigma, mu + 4*sigma, 200)
    y = gestosc(x, mu, sigma)

ax.plot(x, y, color="red", linewidth=2, label=rf"Krzywa gęstości ($\mu$={mu}, $\sigma$={sigma})")
ax.set_title(rf"Rozkład normalny dla $\mu = {mu}$ i $\sigma = {sigma}$")
ax.set_xlabel("Wartość")
ax.set_ylabel("Gęstość")
ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
````
